## Zadanie 1 · Informacje o środowisku

Sprawdź system operacyjny, wersję Pythona, liczbę procesorów, pamięć RAM oraz dostępność GPU.

```python
import platform
import psutil
import shutil
import subprocess

ram = psutil.virtual_memory()
print("=== Informacje o systemie ===")
print(f"System operacyjny: {platform.system()} {platform.release()}")
print(f"Szczegóły systemu: {platform.platform()}")
print(f"Wersja Pythona: {platform.python_version()}")
print(f"Liczba dostępnych CPU: {psutil.cpu_count()}")

print("\n=== Pamięć RAM ===")
print(f"Łącznie: {ram.total / (1024 ** 3):.2f} GB")
print(f"Dostępna: {ram.available / (1024 ** 3):.2f} GB")
print(f"Użycie: {ram.percent}%")

gpu_command = shutil.which("nvidia-smi")
gpu_result = subprocess.run([gpu_command, "--query-gpu=name", "--format=csv,noheader"], capture_output=True, text=True, timeout=10) if gpu_command else None
gpu_names = gpu_result.stdout.strip().splitlines() if gpu_result and gpu_result.returncode == 0 else []
print("\n=== GPU ===")
print("GPU jest dostępne:" if gpu_names else "GPU jest niedostępne.")
print("\n".join(f"- {name}" for name in gpu_names))
```

## Zadanie 2 · Google Drive

Połącz Dysk Google z Colabem, utwórz katalog na pliki z lekcji i zapisz w nim plik z imieniem oraz datą.

```python
from google.colab import drive
from pathlib import Path
from datetime import date

drive.mount('/content/drive')

folder = Path('/content/drive/MyDrive/DS_Course/Lesson02')
folder.mkdir(parents=True, exist_ok=True)

imie = 'Justyna'
data = date.today().isoformat()
plik = folder / f'{imie}_{data}.txt'
plik.write_text(f'Imię: {imie}\nData: {data}\n', encoding='utf-8')

print('Folder istnieje:', folder.is_dir())
print('Zawartość folderu:')
for element in sorted(folder.iterdir()):
    print('-', element.name)
```
